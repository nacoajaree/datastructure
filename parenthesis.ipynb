```python
def check_brackets(expression):
    stack = []
    # สร้างคู่พจนานุกรม (Dictionary) เพื่อจับคู่วงเล็บเปิด-ปิด
    bracket_map = {")": "(", "]": "[", "}": "{"}

    for char in expression:
        # 1. ถ้าเจอวงเล็บเปิด ให้เก็บเข้าสแตก
        if char in ["(", "[", "{"]:
            stack.append(char)

        # 2. ถ้าเจอวงเล็บปิด
        elif char in [")", "]", "}"]:
            # ถ้าเจอวงเล็บปิด แต่ในสแตกไม่มีวงเล็บเปิดเหลืออยู่เลย แปลว่าวงเล็บเกิน/ผิดคู่
            if not stack:
                return False

            # ดึงวงเล็บเปิดตัวล่าสุดออกมาดูว่าตรงคู่กับวงเล็บปิดตัวนี้ไหม
            top_element = stack.pop()
            if bracket_map[char] != top_element:
                return False  # ถ้าไม่ตรงคู่กัน เช่นเจอ ( ] จะถือว่าผิดทันที

    # 3. ตรวจสอบสุดท้ายหลังจากอ่านจบประโยค ถ้าสแตกโล่ง แปลว่าจับคู่ถูกต้องสมบูรณ์
    return len(stack) == 0

# --- ทดลองใช้งาน ---
if __name__ == "__main__":
    # ตัวอย่างเคสที่ถูกต้อง
    test_ok1 = "([]{})"
    test_ok2 = "if (x > 0) { array[0] = 5; }"

    # ตัวอย่างเคสที่ผิด
    test_fail1 = "([)]"      # ผิดลำดับ: วงเล็บเหลี่ยมปิดก่อนวงเล็บกลม
    test_fail2 = "(()"       # ผิดจำนวน: วงเล็บเปิดเกิน
    test_fail3 = "())"       # ผิดจำนวน: วงเล็บปิดเกิน

    print(f"'{test_ok1}'   -> {check_brackets(test_ok1)}")   # True
    print(f"'{test_ok2}'   -> {check_brackets(test_ok2)}")   # True
    print("-" * 35)
    print(f"'{test_fail1}'      -> {check_brackets(test_fail1)}") # False
    print(f"'{test_fail2}'       -> {check_brackets(test_fail2)}") # False
    print(f"'{test_fail3}'       -> {check_brackets(test_fail3)}") # False
```
